```python
import pandas as pd
pd.set_option('display.max_columns', 50)

# Dataset was gathered via Kaggle in this link: https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset
df = pd.read_csv('dataset.csv', index_col = [0])

# convert duration_ms to seconds for more user-friendly metrics
df['duration_s'] = df['duration_ms'] *(1/1000) * (1/60)
```

```python
df.shape
```

### **Data cleaning starts here:**
Removal of invalid rows

```python
# Sanity check of every columns
df.info()
```

```python
# Data sanity check:
# Drop columns with invalid amounts
df = df.drop_duplicates()
df = df.drop(df.query('(duration_ms == 0) or (time_signature == 0) or (tempo == 0)').index)
df = df.drop(['track_id', 'duration_ms'], axis=1)
df.info()
```

```python
# Manual standardization of the value due to extremely low values such as 0.0001
df.loc[df['instrumentalness'] <= 0.0099, 'instrumentalness'] = 0.01
df.loc[df['energy'] <= 0.0099, 'energy'] = 0.01
```
