```python
import pandas as pd

pd.set_option('display.max_columns', 50)

df = pd.read_csv('dataset.csv', index_col = [0])
# Drop track_id column as we don't need it in this analysis/machine learning model

df['duration_s'] = df['duration_ms'] *(1/1000) * (1/60)
df = df.drop_duplicates()
# df = df.drop(['track_id', 'duration_ms'], axis=1)
```

```python
df[df.duplicated(subset=['track_name', 'time_signature'], keep=False)].sort_values('track_name', ascending=False)
```

```python
df.groupby(['track_name', 'time_signature']).agg(bilang = ('track_id', 'count')).sort_values('bilang', ascending=False).reset_index()
```
